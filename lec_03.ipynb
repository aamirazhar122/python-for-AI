"and"	Returns True if both statements are true.	exp:(5 > 3) and (10 == 10) (True)
"or"	Returns True if at least one of the statements is true.	exp:(5 > 3) or (10 == 11) (True)
"not"	Reverses the result; returns False if the result is true, and vice-versa. exp:(5 > 3) (False)
