If the user enters Joke then we will print out a single joke. Each time the joke is always the same.
If the user enters anything else we print out:

```python
# solution from dictionary 

planary_weight_constant : dict = { "Mercury": 37.6, "Venus": 88.9,

"Mars": 37.8, "Jupiter": 236.0, "Saturn": 108.1,"Uranus": 81.5,"Neptune": 114.0
}
def weight_planet():
    weight_on_earth = int(input("Enter your weight: "))
    print(f"weight on earth :{weight_on_earth}")

    choose_planet = str(input("Enter planet name: "))
    print(f"you choose {choose_planet} planet")
    if choose_planet in planary_weight_constant :
        weight_on_planet = weight_on_earth * planary_weight_constant[choose_planet] / 100
        print(f"weight on planet : {weight_on_planet}")
    else:
        print("only calculate weight on planet")    
weight_planet()        
```
