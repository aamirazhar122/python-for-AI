```python
users = {
    "name" : "aamir",
    "age" :  32,
    "profession" : "engineer"
}
print(users["name"])
print(users["age"])
print(users["profession"])
```

```python
#update dictionary
users = {
    "name" : "aamir",
    "age"  :  33,
    "city" : "Islamabad"
}
print(users["city"])

#update
users["city"] = "Faisalabad"
print(users["city"])

users["age"] = 32
print(users["age"])
```

# Loop

```python
for i in range(1 , 4) :
    print(i)
```

```python
#for loop
students : list = ["Ali", "Raza", "Anwar", "saim", "aamir", "azhar"]
for student in students :
    print(student) 
```

```python
#while loop
correct_num = 8
guess = 0

while guess != correct_num:
    guess = int(input("write your number here"))
    print("tyr agin" if guess != correct_num else "ohh, you guessed!")
```

```python
countdown = 10

while countdown >= 1 :
    print(f"{countdown}...")
    countdown -= 1    

print("Happy new year!!!")    
```

```python
#for loop
numbers = [1, 2, 3, 4, 5, 6]
#print square of all numbers
for number in numbers :
    print(number **2)
```
