# file Handling

```python
#open a file
file = open("aamir.txt", "r")
file.close()
```

```python
#write a file
file = open("aamir.txt", "w")
file.write("my elder brother name is Abdul hadi")
file.close()
```

```python
#append the file
file = open("aamir.txt", "a")
file.write("\nmy name is Huda aamir")
file.write("\nI have two(2) brothers")
file.close()
```

```python
#read the file
file = open("aamir.txt", "r")
line = file.read()
print(line)
file.close()
```

```python
#new method: write a file
with open ("aamir.txt", "w+") as file :
    file.write("Hello world")
    file.seek(0)
    print(file.read())
```

```python
# append the file
with open ("aamir.txt", "a+") as newfile:
    newfile.write("\nhello everyone")
    newfile.seek(0)
    read = newfile.read()
    print(read)
```
