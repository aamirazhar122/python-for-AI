```python
#class:
class House:
    area : int = 1250
    address : str = "H #123, Fsd"
    Num_of_rooms : int = 8 

My_house = House()   

print(My_house.address)
print(My_house.Num_of_rooms)
print(My_house.area)
```

```python
class House:
    area : int = 1250
    address : str 
    Num_of_rooms : int = 8 
    def __init__(self, address : str):
        self.address = address
aamir_house = House("H # 456, Fsd")
ali_house = House("H # 321 Lhr")   

print(aamir_house.address)
print(ali_house.Num_of_rooms)
print(ali_house.address )
```

```python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def avg_marks(self) :
        sum = 0
        for val in self.marks:
            sum += val
        print(f"hi,", self.name,", your average marks are :", sum/3)

s1 = Student("aamir", [90, 56, 65])        
s1.avg_marks()
```
