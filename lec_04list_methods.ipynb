# LIST


```python

shopping_items : list[str] = ["cake", "eggs", "milk", "juice"]
print(shopping_items)
```

```python
#                     index#     0      1       2       3
shopping_items : list[str] = ["cake", "eggs", "milk", "juice"]
print(shopping_items[3])
```

```python
shopping_items : list[str] = ["a","b", "cake", "eggs", "milk", "juice"]
print(len(shopping_items)) # length of list = 6
```

```python
shopping_items : list[str] = ["cake", "eggs", "milk", "juice"]
print(shopping_items[0:3])
```

```python
shopping_items : list[str] = ["cake", "eggs", "milk", "juice"]
print("eggs" in shopping_items) # in list
print("bread" in shopping_items) # not in list
```

```python
shopping_items : list[str] = ["a","b", "cake", "eggs", "milk", "juice"]
print(shopping_items)

shopping_items.append("tea")
print(shopping_items)

shopping_items.remove("juice")
print(shopping_items)
```

```python
shopping_items : list[str] = ["a","b", "cake", "eggs", "milk", "juice"]
print(shopping_items)

shopping_items.insert(3, "tea")
print(shopping_items)

shopping_items.pop(0) #poping list
print(shopping_items)
```
