```python
#temperature_status(temp)
def temperature_status(temp):
    if temp > 30 :
        return "hot"
    elif temp == 20 or temp > 20 :
        return "warm"
    elif temp < 20:
        return "cold"
    else :
        return ' '
temperature_status(31)   
```
