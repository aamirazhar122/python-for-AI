# Error Handinhg

```python
try:
    number = int(input("enter your number"))
    result = number / 100
    print(f"result is : {result}")
except ZeroDivisionError: # 2 / 0 = error
    print("cnnot devided by zero")
except ValueError: # n / 10 # error
    print("only numbers are accepted")    
except Exception as e:
    print("some thing happened. Error {e}")
```
